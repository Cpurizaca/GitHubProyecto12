```python
dbutils.widgets.removeAll()
```

```python
from pyspark.sql.functions import *
from pyspark.sql.types import *
from pyspark.sql import functions as F
```

```python
dbutils.widgets.text("catalogo", "catalog_movies")
dbutils.widgets.text("esquema_source", "silver")
dbutils.widgets.text("esquema_sink", "golden")
```

```python
catalogo = dbutils.widgets.get("catalogo")
esquema_source = dbutils.widgets.get("esquema_source")
esquema_sink = dbutils.widgets.get("esquema_sink")
```

```python
df_movies_transformed = spark.table(f"{catalogo}.{esquema_source}.movies_transformed")
```

```python
df_transformed = df_movies_transformed.select(*[col for col in df_movies_transformed.columns[:-1]])
```

```python
df_transformed.write.mode("overwrite").insertInto(f"{catalogo}.{esquema_sink}.golden_movies")
```
