```python
from pyspark.sql.types import StructType, StructField, IntegerType, StringType, DateType
from pyspark.sql.functions import current_timestamp, to_timestamp, concat, col, lit
```

```python
dbutils.widgets.text("container", "raw")
dbutils.widgets.text("catalogo", "catalog_movies")
dbutils.widgets.text("esquema", "bronze")
dbutils.widgets.text("storageName", "adlproyecto12")
```

```python
container = dbutils.widgets.get("container")
catalogo = dbutils.widgets.get("catalogo")
esquema = dbutils.widgets.get("esquema")
storageName = dbutils.widgets.get("storageName")

ruta = f"abfss://{container}@{storageName}.dfs.core.windows.net/Movies.csv"
```

```python
movies_schema = StructType(fields=[StructField("id", IntegerType(), False),
                                  StructField("title", StringType(), True),
                                  StructField("genres", StringType(), True),
                                  StructField("language", StringType(), True),
                                  StructField("user_score", FloatType(), True),
                                  StructField("runtime_hour", IntegerType(), True),
                                  StructField("runtime_min", IntegerType(), True),
                                  StructField("release_date", DateType(), True),
                                  StructField("vote_count", IntegerType(), True),
])
```

```python
movies_df = spark.read \
            .option("header", True) \
            .schema(movies_schema) \
            .csv(ruta)
```

```python
movies_with_timestamp_df = movies_df.withColumn("ingestion_date", current_timestamp())
```

```python
movies_selected_df = movies_with_timestamp_df.select(col('id').alias('movie_id'),
                                                     col('title'),
                                                     col('genres'),
                                                     col('language'),
                                                     col('user_score'),
                                                     col('runtime_hour'),
                                                     col('runtime_min'),
                                                     col('release_date'),
                                                     col('vote_count'),
                                                     col('ingestion_date')
)
```

```python
movies_selected_df.write.mode('overwrite').insertInto(f"{catalogo}.{esquema}.movies")
```
